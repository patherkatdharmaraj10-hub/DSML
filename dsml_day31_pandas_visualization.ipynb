Crosstab tells us how many observations belong to each combination of categories.

```python
pd.crosstab(df["survived"], df["sex"]).plot(
#df["survived"].value_counts().plot(
kind = 'bar',
title = "Count of survived passengers",
xlabel="No. of passengers",
ylabel= "Survived (0 = No, 1 = yes)",
figsize= (10,9),
legend = True,
#yticks = list(range(0,600,50)),
#color = 'red',
grid = True,
colormap = "Dark2"
)
```

```python


# Count the number of passengers from each embarkation location
embarked_counts = df["embarked"].value_counts()

# Create the pie chart
embarked_counts.plot(
    kind="pie",
    labels=embarked_counts.index,
    autopct="%1.1f%%",
    startangle=0,
    figsize=(8, 6),
    legend=True
)

# Add title
plt.title("Passengers by Embarked Location")

# Add Y-axis label
plt.ylabel("Age")

# Display the chart
plt.show()
```

```python
# Count non-missing Age values for each Embarked category
age_count = df.groupby("embarked")["age"].count()

# Create pie chart
age_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90,
    figsize=(8, 6)
)

# Add title
plt.title("Age Count by Embarked")

# Remove y-axis label
plt.ylabel("")

# Display the pie chart
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize = (10, 5))

female_data = df[df["sex"] =="female"]["survived"].value_counts()

female_data.plot(
    kind = "pie",
    ax = axes[0],
    labels = female_data.index,
    autopct= None,
    legend = True
)

axes[0].set_ylabel("female")

male_data = df[df["sex"] == "male"]["survived"].value_counts()

male_data.plot(
    kind = "pie",
    ax = axes[1],
    labels = male_data.index,
    autopct = None,
    legend = True,
)

axes[1].set_ylabel("male")

# adust layout
plt.tight_layout()

plt.show()
```

### Question 2: Heatmap of Correlations between Numerical Variables
what is correlation?
correlation measures how strongly two numerical variables are related.
The value is between
-1 and +1

for example:
  +1 -> very strong positive relationship
   0 -> little/no linear relationship
   -> very strong negative relationship

Select numerical columns

```python
numerical_df = df.select_dtypes(include = "number")
numerical_df.head()
```

Calculate correlation

```python
correlation = numerical_df.corr()
correlation
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    correlation,
    annot = True,
    cmap = "coolwarm",
    fmt = '.2f'
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.show()
```

### Question 3: Compare Survival Count with and without Siblings/spouse

```python
df["SibSp_Status"] = np.where(
    df['sibsp'] > 0,
    'with Siblings/Spouse',
    'without Siblings/Spouse'
)
```

Create readable survival labels

```python
df['Survived_Status'] = df['survived'].map({
    0: 'Did not Survive',
    1: "Did not Survive"
})
```

Count plot

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    x='SibSp_Status',
    hue='Survived_Status',
    data=df
)

plt.title('Survival Count With and Without Siblings/Spouse')
plt.xlabel('Siblings/Spouse Status')
plt.ylabel('Number of Passengers')

plt.show()
```

### Question 4: Violin Plot of age vs Survival
** A violit plot shows the distribution of a numerical variable across categories.**

here,
age is the numerical vaiable and survival is category

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    x='Survived_Status',
    y='age',
    data=df
)

plt.title('Age Distribution by Survival Status')
plt.xlabel('Survival Status')
plt.ylabel('Age')

plt.show()
```
