What is correlation?

Correlation measures the relationship between two numerical variables.

Its value is between -1 and +1:

Correlation	Meaning
+1	Perfect positive relationship
+0.7	Strong positive relationship
+0.3	Weak positive relationship
0	No linear relationship
-0.3	Weak negative relationship
-0.7	Strong negative relationship
-1	Perfect negative relationship

### What is a heatmap?

### ScatterPlot
📈 Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

For example, with animals:

X-axis → animal weight
Y-axis → animal height
Each dot → one animal

If the dots generally move upward from left to right, there is a positive relationship.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    df["age"],
    df["fare"],
    alpha=0.5
)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

### Measure of Central Tendancy

Measures of Central Tendency

Measures of central tendency are statistical methods used to find the central or typical value of a dataset.

The three main measures are:

Mean
Median
Mode

Think of them as answering:

"What is the typical value in my data?"


```python
from scipy import stats
```

```python
x = np.random.standard_t(500, size=1000)
```

```python
print(x)
```

```python
_mean = np.mean(x)
```

```python
_mean
```

```python
plt.hist(x, bins =20, color = "green")
```

```python
plt.hist(x, bins = 20, color = "blue")
plt.axvline(x=_mean, color ="red")
```

```python
x = np.random.standard_t(500, size=1000)
median = np.median(x)
```

```python
median
```

```python
plt.hist(x, bins = 10, color= "black")
plt.axvline(x = median, color = "blue")
```

```python
mode = stats.mode(x)
```

```python
mode
```

```python

mode = stats.mode(x, keepdims=False).mode

print(mode)
print(type(mode))
```

```python
plt.hist(x, bins=10, color="red")

plt.axvline(x=mode, color="black", linewidth=2)

plt.show()
```
